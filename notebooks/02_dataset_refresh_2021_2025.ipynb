## Schema Mapping

The original K-drama dataset and the new 2021–2025 dataset use different column names and slightly different structures. To keep the existing Streamlit project intact, both datasets will be standardized into one common schema before being merged.

| Final Column Name | Old Dataset Column | New Dataset Column | Notes |
|---|---|---|---|
| title | Name | title | Main drama title |
| normalized_title | generated | generated | Used for fuzzy matching and deduplication |
| start_year | Year | startYear | Drama start year |
| end_year | Year | endYear | Old dataset has only one year, so end_year will use Year |
| genres | Genre | genres | Genre/category information |
| cast | Main Cast | mainLead1 + mainLead2 | New dataset only has two lead actors |
| main_lead_1 | missing | mainLead1 | First lead actor |
| main_lead_2 | missing | mainLead2 | Second lead actor |
| synopsis | Sinopsis | synopsis | Drama description |
| global_rating | Score | rating | Public/global rating |
| content_rating | Content Rating | missing | Available only in old dataset |
| tags | Tags | missing | Available only in old dataset |
| platform | Network | missing | New dataset will use "Unknown" |
| poster_url | img url | image_url | Poster/image link |
| episodes | Episode | missing | Available only in old dataset |
| votes | missing | votes | Available only in new dataset |
| writer | missing | writer | Available only in new dataset |
| director | missing | director | Available only in new dataset |
| source_dataset | generated | generated | Tracks source file |

```python
import re
import numpy as np

def normalize_title(title):
    """
    Standardizes drama titles so fuzzy matching and deduplication work better.
    Example:
    'Crash Landing on You!' -> 'crash landing on you'
    """
    if pd.isna(title):
        return ""
    
    title = str(title).lower()
    title = re.sub(r"[^a-z0-9\s]", "", title)
    title = re.sub(r"\s+", " ", title).strip()
    
    return title
```

Standardize the old dataset

```python
old_clean = pd.DataFrame()

old_clean["title"] = old_df["Name"]
old_clean["normalized_title"] = old_clean["title"].apply(normalize_title)

old_clean["start_year"] = pd.to_numeric(old_df["Year"], errors="coerce")
old_clean["end_year"] = pd.to_numeric(old_df["Year"], errors="coerce")

old_clean["genres"] = old_df["Genre"]
old_clean["cast"] = old_df["Main Cast"]

old_clean["main_lead_1"] = np.nan
old_clean["main_lead_2"] = np.nan

old_clean["synopsis"] = old_df["Sinopsis"]
old_clean["global_rating"] = pd.to_numeric(old_df["Score"], errors="coerce")

old_clean["content_rating"] = old_df["Content Rating"]
old_clean["tags"] = old_df["Tags"]
old_clean["platform"] = old_df["Network"]
old_clean["poster_url"] = old_df["img url"]
old_clean["episodes"] = old_df["Episode"]

old_clean["votes"] = np.nan
old_clean["writer"] = np.nan
old_clean["director"] = np.nan

old_clean["source_dataset"] = "original_1500"

old_clean.head()
```

Standardize the new dataset

```python
new_clean = pd.DataFrame()

new_clean["title"] = new_df["title"]
new_clean["normalized_title"] = new_clean["title"].apply(normalize_title)

new_clean["start_year"] = pd.to_datetime(
    new_df["startYear"], errors="coerce"
).dt.year

new_clean["end_year"] = pd.to_datetime(
    new_df["endYear"], errors="coerce"
).dt.year

new_clean["genres"] = new_df["genres"]

new_clean["main_lead_1"] = new_df["mainLead1"]
new_clean["main_lead_2"] = new_df["mainLead2"]

new_clean["cast"] = (
    new_df["mainLead1"].fillna("").astype(str)
    + ", "
    + new_df["mainLead2"].fillna("").astype(str)
)

new_clean["cast"] = new_clean["cast"].str.strip(", ")

new_clean["synopsis"] = new_df["synopsis"]
new_clean["global_rating"] = pd.to_numeric(new_df["rating"], errors="coerce")

new_clean["content_rating"] = np.nan
new_clean["tags"] = np.nan

# Platform is missing in the new dataset, so we mark it honestly as Unknown
new_clean["platform"] = "Unknown"

new_clean["poster_url"] = new_df["image_url"]
new_clean["episodes"] = np.nan

new_clean["votes"] = pd.to_numeric(new_df["votes"], errors="coerce")
new_clean["writer"] = new_df["writer"]
new_clean["director"] = new_df["director"]

new_clean["source_dataset"] = "kaggle_2021_2025"

new_clean.head()
```

```python
final_columns = [
    "title",
    "normalized_title",
    "start_year",
    "end_year",
    "genres",
    "cast",
    "main_lead_1",
    "main_lead_2",
    "synopsis",
    "global_rating",
    "content_rating",
    "tags",
    "platform",
    "poster_url",
    "episodes",
    "votes",
    "writer",
    "director",
    "source_dataset"
]

old_clean = old_clean[final_columns]
new_clean = new_clean[final_columns]
```

### Combine both datasets

```python
def is_valid_value(value, treat_unknown_as_missing=True):
    """
    Checks whether a value is useful.
    Treats blanks, NaN, None, and optionally 'Unknown' as missing.
    """
    if pd.isna(value):
        return False
    
    if isinstance(value, str):
        cleaned = value.strip()
        if cleaned == "":
            return False
        
        missing_strings = {"nan", "none", "null", "n/a", "na"}
        
        if treat_unknown_as_missing:
            missing_strings.add("unknown")
        
        if cleaned.lower() in missing_strings:
            return False
    
    return True


def first_valid_from(group, column, source=None, treat_unknown_as_missing=True):
    """
    Gets the first useful value from a specific source if provided.
    """
    if source:
        group = group[group["source_dataset"] == source]
    
    for value in group[column]:
        if is_valid_value(value, treat_unknown_as_missing=treat_unknown_as_missing):
            return value
    
    return np.nan


def longest_valid_text(group, column):
    """
    For fields like synopsis, keeps the longest available text.
    """
    valid_values = [
        value for value in group[column]
        if is_valid_value(value)
    ]
    
    if not valid_values:
        return np.nan
    
    return max(valid_values, key=lambda x: len(str(x)))
```

```python
def merge_drama_group(group):
    """
    Merges duplicate drama rows into one best possible row.
    
    Priority logic:
    - Keep old dataset values for platform, year, tags, content rating, episodes
    - Keep new dataset values for votes, writer, director, and main leads
    - Keep the longest synopsis
    - Keep original rating when available for consistency with the old project
    """
    
    row = {}
    
    old_source = "original_1500"
    new_source = "kaggle_2021_2025"
    
    # Basic identity fields
    row["title"] = first_valid_from(group, "title", old_source)
    if not is_valid_value(row["title"]):
        row["title"] = first_valid_from(group, "title", new_source)
    
    row["normalized_title"] = first_valid_from(group, "normalized_title", old_source)
    if not is_valid_value(row["normalized_title"]):
        row["normalized_title"] = first_valid_from(group, "normalized_title", new_source)
    
    # Year fields: prefer old dataset because your output shows new start_year is often NaN
    row["start_year"] = first_valid_from(group, "start_year", old_source)
    if not is_valid_value(row["start_year"]):
        row["start_year"] = first_valid_from(group, "start_year", new_source)
    
    row["end_year"] = first_valid_from(group, "end_year", old_source)
    if not is_valid_value(row["end_year"]):
        row["end_year"] = first_valid_from(group, "end_year", new_source)
    
    # Drama details
    row["genres"] = first_valid_from(group, "genres", old_source)
    if not is_valid_value(row["genres"]):
        row["genres"] = first_valid_from(group, "genres", new_source)
    
    # Old cast is usually richer; new has only mainLead1 and mainLead2
    row["cast"] = first_valid_from(group, "cast", old_source)
    if not is_valid_value(row["cast"]):
        row["cast"] = first_valid_from(group, "cast", new_source)
    
    # New dataset has structured lead columns
    row["main_lead_1"] = first_valid_from(group, "main_lead_1", new_source)
    row["main_lead_2"] = first_valid_from(group, "main_lead_2", new_source)
    
    # Keep longest available synopsis
    row["synopsis"] = longest_valid_text(group, "synopsis")
    
    # Prefer old rating for consistency with existing app
    row["global_rating"] = first_valid_from(group, "global_rating", old_source)
    if not is_valid_value(row["global_rating"]):
        row["global_rating"] = first_valid_from(group, "global_rating", new_source)
    
    # Old-only app fields
    row["content_rating"] = first_valid_from(group, "content_rating", old_source)
    row["tags"] = first_valid_from(group, "tags", old_source)
    
    # Platform: prefer old because new dataset has Unknown
    row["platform"] = first_valid_from(group, "platform", old_source)
    if not is_valid_value(row["platform"]):
        row["platform"] = "Unknown"
    
    # Poster: prefer old MyDramaList poster if available, otherwise use new image_url
    row["poster_url"] = first_valid_from(group, "poster_url", old_source)
    if not is_valid_value(row["poster_url"]):
        row["poster_url"] = first_valid_from(group, "poster_url", new_source)
    
    row["episodes"] = first_valid_from(group, "episodes", old_source)
    
    # New-only useful future ML fields
    row["votes"] = first_valid_from(group, "votes", new_source)
    row["writer"] = first_valid_from(group, "writer", new_source)
    row["director"] = first_valid_from(group, "director", new_source)
    
    # Track whether row came from one or both datasets
    sources = sorted(group["source_dataset"].dropna().unique())
    row["source_dataset"] = " + ".join(sources)
    
    return row
```

```python
merged_rows = []

for normalized_title, group in combined_df.groupby("normalized_title", dropna=False):
    merged_rows.append(merge_drama_group(group))

master_df = pd.DataFrame(merged_rows)

# Keep final column order
master_df = master_df[final_columns]

print("Combined dataset before deduplication:", combined_df.shape)
print("Master dataset after smart deduplication:", master_df.shape)
print("Rows removed:", combined_df.shape[0] - master_df.shape[0])
print("Remaining duplicate normalized titles:", master_df.duplicated(subset=["normalized_title"]).sum())
```

## Create Platform Override File

The 2021–2025 dataset does not include streaming platform information. To avoid hardcoding platform values in the app, a separate `platform_overrides.csv` file is created. This file can be manually updated over time and later merged into the app through the data loader.

```python
# Create a starter platform override file for dramas with Unknown platform

platform_override_path = DATA_DIR / "platform_overrides.csv"

unknown_platform_df = master_df[
    master_df["platform"].fillna("Unknown").str.strip().str.lower() == "unknown"
][["title", "normalized_title"]].copy()

unknown_platform_df["platform"] = ""
unknown_platform_df["notes"] = "Manual platform lookup needed"

unknown_platform_df.to_csv(platform_override_path, index=False, encoding="utf-8")

print(f"Platform override file created at: {platform_override_path}")
print("Number of dramas needing platform lookup:", unknown_platform_df.shape[0])

unknown_platform_df.head(20)
```
